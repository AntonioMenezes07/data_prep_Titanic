##**importando a base de dados**

```python
import pandas as pd
df = pd.read_csv('/content/train.csv')
df
```

```python
lista_spec = ['PassengerId', 'Survived']
```

```python
df01 = df.drop(lista_spec,axis = 1)
df01
```

##**extraindo titulo dos nomes**

```python
df01['titulo'] = df01['Name'].str.extract(' (Mr\.|Miss\.|Mrs\.|Dr\.) ')
```

```python
df01 = df01.drop('Name',axis = 1)
df01
```

##**Metadados**

```python
metadados = pd.DataFrame({'colunas': df01.columns, 'tipo': df01.dtypes, 'cardinalidade': df01.nunique(),'qtd_nulos' : df01.isnull().sum() ,'percent_nulos': round((df01.isnull().sum()/len(df01))*100,2)})
metadados=metadados.sort_values(by='percent_nulos',ascending=False)
metadados = metadados.reset_index(drop=True)
metadados
```

##**dropando colunas com mais de 75% de valores nulos**

```python
df_drop_nulos = metadados[metadados['percent_nulos']>= 75]
lista_drop_nulos = list(df_drop_nulos['colunas'])
lista_drop_nulos
```

```python
df01 = df01.drop(df_drop_nulos['colunas'], axis = 1)
df01
```

##**tratamento de valores nulos**

####**variaveis numericas**
trocar pela media

```python
df01 = df01.fillna(df01.select_dtypes(include=['int64', 'float64']).mean())
df01
```

####**variaveis categoricas**
trocar por 'desconhecido'

```python
df01[df01.select_dtypes(include=['object']).columns] = df01[df01.select_dtypes(include=['object']).columns].fillna('desconhecido')
df01
```

##**normalizacao dos dados**
z-score

```python
from scipy.stats import zscore
df01[df01.select_dtypes(include = ['int64', 'float64']).columns] = df01[df01.select_dtypes(include = ['int64', 'float64']).columns].apply(zscore)
df01
```

##**aplicando o oneHotEnconding para variaveis categoricas com cardinalidade baixa**

```python
from sklearn.preprocessing import OneHotEncoder
cardinalidade = 2
variaveis = metadados[(metadados['cardinalidade'] <= cardinalidade) & (metadados['tipo'] == 'object')]
variaveis = list(variaveis['colunas'])
variaveis
```

```python
# Instanciando o encoder
encoder = OneHotEncoder(drop='first', sparse_output=False)

# Aplicando o one-hot encoding
encoded_data = encoder.fit_transform(df01[variaveis])

encoded_cols = encoder.get_feature_names_out(variaveis)
encoded_df = pd.DataFrame(encoded_data, columns=encoded_cols, index=df01.index)

df01 = pd.concat([df01.drop(variaveis, axis=1), encoded_df], axis=1)

df01.head()
```

##**aplicando o label enconding para variaveis com alta cardinalidade**

```python
cardinalidade = 2
variaveis02 = metadados[(metadados['cardinalidade'] > cardinalidade) & (metadados['tipo'] == 'object')]
variaveis02 = list(variaveis02['colunas'])
variaveis02


for item in lista_drop_nulos:
    if item in variaveis02:
        variaveis02.remove(item)
variaveis02
```

```python
from sklearn.preprocessing import LabelEncoder

for col in variaveis02:
    encoder = LabelEncoder()
    df01[col] = encoder.fit_transform(df01[col])

df01.head()

# Inner join usando o índice
df02 = pd.merge(df01, df[['Survived']], left_index=True, right_index=True, how='inner')
df02.shape
```
